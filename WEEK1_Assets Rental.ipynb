```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
# we can give an index number or name for our index column, or leave it blank
df = pd.read_csv('/Users/prakashghatani/Desktop/Regis/MSDS692_Practicum 1/WEEK1/data/Jan LAR.csv')
df
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Convert date columns
df["Commencement Date"] = pd.to_datetime(df["Commencement Date"], errors="coerce")
df["Primary Term Date"] = pd.to_datetime(df["Primary Term Date"], errors="coerce")

# Basic check
df.head()
```

```python
missing = (
    df.isna().mean().mul(100)
      .sort_values(ascending=False)
      .head(12)
      .reset_index()
)
missing.columns = ['Column', 'Missing Percent']

display(missing)

fig_missing = px.bar(
    missing,
    x='Missing Percent',
    y='Column',
    orientation='h',
    title='Top Missing Data Fields',
    text='Missing Percent'
)
fig_missing.update_traces(texttemplate='%{text:.1f}%', textposition='outside')
fig_missing.update_layout(yaxis={'categoryorder': 'total ascending'}, height=500)
fig_missing.show()
```

```python
# Top asset descriptions
top_assets = df["Asset Description"].value_counts().head(10).reset_index()
top_assets.columns = ["Asset Description", "Count"]

fig = px.bar(
    top_assets,
    x="Asset Description",
    y="Count",
    title="Top 10 Asset Descriptions"
)
fig.show()
```

```python
# Asset state distribution
state_counts = df["Asset State"].value_counts().reset_index()
state_counts.columns = ["Asset State", "Count"]

fig = px.pie(
    state_counts,
    names="Asset State",
    values="Count",
    title="Asset Distribution by State"
)
fig.show()
```

```python
# Contract type distribution
contract_type = df["Contract Type"].value_counts().reset_index()
contract_type.columns = ["Contract Type", "Count"]

fig = px.bar(
    contract_type,
    x="Contract Type",
    y="Count",
    title="Contract Type Distribution"
)
fig.show()
```

```python
# Asset Cost vs Asset Rental Amount
fig = px.scatter(
    df,
    x="Asset Cost",
    y="Asset Rental Amount",
    color="Asset Description",
    hover_data=["Contract Number", "Service Tag"],
    title="Asset Cost vs Asset Rental Amount"
)
fig.show()
```

```python
# Correlation heatmap
numeric_df = df.select_dtypes(include=["int64", "float64"])
corr = numeric_df.corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="Correlation Heatmap"
)
fig.show()
```

```python
# Contract-level summary
contract_summary = df.groupby("Contract Number").agg(
    Asset_Count=("Asset Number", "count"),
    Total_Asset_Cost=("Asset Cost", "sum"),
    Total_Monthly_Rent=("Asset Rental Amount", "sum"),
    Avg_Asset_Cost=("Asset Cost", "mean"),
    End_Date=("Primary Term Date", "max")
).reset_index()

top_contracts = contract_summary.sort_values(
    "Total_Monthly_Rent",
    ascending=False
).head(10)

fig = px.bar(
    top_contracts,
    x="Contract Number",
    y="Total_Monthly_Rent",
    hover_data=["Asset_Count", "Total_Asset_Cost", "End_Date"],
    title="Top 10 Contracts by Monthly Rental Amount"
)
fig.show()
```

```python
# Rental rate feature
df["Rental_Rate"] = df["Asset Rental Amount"] / df["Asset Cost"]

top_rental_rate = df.sort_values("Rental_Rate", ascending=False).head(10)

fig = px.bar(
    top_rental_rate,
    x="Service Tag",
    y="Rental_Rate",
    hover_data=["Asset Description", "Asset Cost", "Asset Rental Amount"],
    title="Top 10 Assets by Rental Rate"
)
fig.show()
```

```python
# Contract status
today = pd.to_datetime("2026-05-16")

df["Contract_Status"] = np.where(
    df["Primary Term Date"] < today,
    "Expired",
    "Active"
)

status_counts = df["Contract_Status"].value_counts().reset_index()
status_counts.columns = ["Contract Status", "Count"]

fig = px.bar(
    status_counts,
    x="Contract Status",
    y="Count",
    title="Active vs Expired Contracts"
)
fig.show()
```

```python
# Contract end year analysis
df["End_Year"] = df["Primary Term Date"].dt.year

end_year_summary = df.groupby("End_Year").agg(
    Asset_Count=("Asset Number", "count"),
    Total_Monthly_Rent=("Asset Rental Amount", "sum")
).reset_index()

fig = px.bar(
    end_year_summary,
    x="End_Year",
    y="Total_Monthly_Rent",
    hover_data=["Asset_Count"],
    title="Monthly Rental Amount by Contract End Year"
)
fig.show()
```

```python
# Boxplot of rental amount by asset description
top_asset_names = df["Asset Description"].value_counts().head(8).index
top_asset_df = df[df["Asset Description"].isin(top_asset_names)]

fig = px.box(
    top_asset_df,
    x="Asset Description",
    y="Asset Rental Amount",
    title="Rental Amount Distribution by Top Asset Types"
)
fig.show()
```

```python
# Simple financial summary
print("Total Asset Cost:", df["Asset Cost"].sum())
print("Total Monthly Rental Amount:", df["Asset Rental Amount"].sum())
print("Annual Rental Amount:", df["Asset Rental Amount"].sum() * 12)
print("Average Asset Cost:", df["Asset Cost"].mean())
print("Average Asset Rental Amount:", df["Asset Rental Amount"].mean())
```
